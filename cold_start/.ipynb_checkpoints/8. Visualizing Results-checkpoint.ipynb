```python
import os
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
from matplotlib.lines import Line2D
import pandas as pd

# Configure font sizes.
TICK_SIZE = 6
SMALL_SIZE = 6
MEDIUM_SIZE = 8
BIGGER_SIZE = 11
plt.rc("font", size=SMALL_SIZE)  # controls default text sizes
plt.rc("axes", titlesize=MEDIUM_SIZE)  # fontsize of the axes title
plt.rc("axes", labelsize=MEDIUM_SIZE)  # fontsize of the x and y labels
plt.rc("xtick", labelsize=TICK_SIZE)  # fontsize of the tick labels
plt.rc("ytick", labelsize=TICK_SIZE)  # fontsize of the tick labels
plt.rc("legend", fontsize=SMALL_SIZE)  # legend fontsize
plt.rc("figure", titlesize=BIGGER_SIZE)  # fontsize of the figure title

MARKERSIZE = 3
LINEWIDTH = 0.5
plt.rc("lines", markersize=MARKERSIZE, linewidth=LINEWIDTH)
plt.rc("grid", linewidth=0.5)
plt.rc("grid", alpha=0.5)

# colorblind friendly colors: https://gist.github.com/thriveth/8560036
colors = ["#377eb8", "#ff7f00", "#4daf4a", "#f781bf", "#a65628", "#984ea3", "#999999", "#e41a1c", "#dede00"]

# load in our files
faq_summary = pd.read_csv("logs/gpqa-all/faq_transfer_0_summary.csv")
best_baseline_summary = pd.read_csv("logs/gpqa-all/best_baseline_summary.csv")
uniform_summary = pd.read_csv("logs/gpqa-all/uniform_summary.csv")
```

# Skinny Main Text Bars

```python
# just need skinny grouped bar plots (note N_q = 1192)
fig, ax = plt.subplots(dpi=400, figsize=(5.5, 1.4)); N_q = 1192
width = 10; N_QUESTIONS = 1192

# plot the FAQ results
ax.bar(
    x=(faq_summary.prop_budget * N_QUESTIONS) - (width // 2), 
    height=faq_summary.ess_multiplier, width=width, 
    yerr=faq_summary.ess_multiplier_serr, capsize=width * 0.4, color=colors[0], 
    label="FAQ")

# plot the best post-hoc baseline
ax.bar(
    x=(baseline_summary.prop_budget * N_QUESTIONS) + (width // 2), 
    height=baseline_summary.ess_multiplier, width=width, 
    yerr=baseline_summary.ess_multiplier_serr, capsize=width * 0.4, color=colors[1], 
    label="Best Baseline (Post-hoc Per Budget)")

# horizontal line at y=1.0 (uniform)
ax.axhline(y=1.0, color="red", linestyle="--", label="Uniform")
ax.set_ylim(bottom=0.8, top=1.4); ax.grid()

# beautify
ax.set_xticks((baseline_summary.prop_budget * N_QUESTIONS).astype(int))
ax.set_xlabel("Budget (Out of 1,192 Total Questions)")
ax.set_ylabel("ESS Multiplier")
ax.legend(ncol=3)
plt.tight_layout()
plt.savefig("figures/5_cold_start_gpqa_main_text.pdf", facecolor="white", bbox_inches="tight")
plt.show()
```

# Coverage Figure in Appendix

```python
# just need skinny line plot for coverage (note N_q = 1192)
fig, ax = plt.subplots(dpi=400, figsize=(5.5, 1.4)); N_q = 1192
width = 10; N_QUESTIONS = 1192

# get our data
q_faq = faq_summary
q_baseline = best_baseline_summary
q_uniform = uniform_summary

# FAQ
ax.errorbar(
    q_faq.prop_budget * N_QUESTIONS, 
    q_faq.coverage, yerr=q_faq.coverage_serr, 
    marker="o", capsize=MARKERSIZE, capthick=1.0, color=colors[0], 
    label="FAQ")

# Baseline
ax.errorbar(
    q_baseline.prop_budget * N_QUESTIONS, 
    q_baseline.coverage, yerr=q_baseline.coverage_serr, 
    marker="x", capsize=MARKERSIZE, capthick=1.0, color=colors[1], 
    label="Best Baseline (Post-hoc Per Budget)")

# Uniform
ax.errorbar(
    q_uniform.prop_budget * N_QUESTIONS, 
    q_uniform.coverage, yerr=q_uniform.coverage_serr, 
    marker="^", capsize=MARKERSIZE, capthick=1.0, color=colors[2], 
    label="Uniform")

# put the grids in + show a reasonable ylim
ax.grid(); ax.set_ylabel("Coverage")
ax.set_ylim(bottom=0.80, top=1.0)
ax.axhline(y=0.95, color="black", linestyle="--", label="95% Coverage")
ax.set_xlabel(f"Budget (Out of 1,192 Total Questions)")

plt.tight_layout(); plt.legend(ncol=2)
plt.savefig("figures/5_cold_start_gpqa_appendix_coverage.pdf", facecolor="white", bbox_inches="tight")
plt.show()
```
